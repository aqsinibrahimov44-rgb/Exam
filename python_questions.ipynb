#### 3. Verilmiş listdə hər sözün neçə dəfə təkrarlandığını dictionary şəklində qaytarın. `Counter` istifadə etməyin.
###### Misal: words = ["apple", "banana", "apple", "orange", "banana", "apple"]
###### Cavab: {'apple': 3, 'banana': 2, 'orange': 1}


```python
words =["apple", "banana", "apple", "orange", "banana", "apple"]
caunt = {}
for i in words:
    if i in caunt:
        caunt[i] += 1
    else:
        caunt[i] = 1
caunt
```

#### 4. List comprehension istifadə edərək Selsi ilə verilmiş temperaturlardan yalnız müsbət olanları Farengeytə çevirin. Düstur: F = C × 9/5 + 32 
###### Misal: celsius = [-5, 0, 10, 20, 30, -10, 15]
###### Cavab: [50.0, 68.0, 86.0, 59.0]


```python
celsius = [-5, 0, 10, 20, 30, -10, 15]
celsius = [temp for temp in celsius if temp > 0]
farangeit =[((9/5)*temp+32) for temp in celsius]
print(farangeit)
```

#### 6. Verilmiş listdən boş (`None`) dəyərləri silin və qalan ədədlərin ortasını tapın. 
###### Misal: data = [23, None, 18, 30, None, 25]
###### Cavab: [23, 18, 30, 25]   Orta: 24.0


```python
data =[23,None,18,30,None,25]
data = [x for x in data if x is not None]
print(data)
avg = sum(data) / len(data)
print(avg)
```

#### 12. `Dataset` adlı class yaradın. Class ad (`name`) və ədədlər listi (`values`) qəbul etsin. Aşağıdakı metodları yazın: `size()` dəyərlərin sayını, `mean()` isə ortasını qaytarsın. 
###### Misal: ds = Dataset('temperatur', [10, 20, 30, 40, 50])     ds.size(), ds.mean()
###### Cavab: 5   30.0
###### İpucu: dəyərləri `__init__`-də `self.values`-a yazın

```python
class Dataset:
    def __init__(self,name,values):
        self.name=name
        self.values=values

temp1_dataset = Dataset("Temperature", [10, 20, 30, 40, 50])
size = len(temp1_dataset.values)
print(size)
mean = sum(temp1_dataset.values) / size
print(mean)
```
